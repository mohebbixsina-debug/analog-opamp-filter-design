```python
import numpy as np
N = 11
# Given specs
A_max_db = 1          # Amax = 1 dB
A_min_db = 25         # Amin = 25 dB
f_p = 14e3            # fp = 14 kHz
f_s = 20e3            # fs = 20 kHz
# GAIN = 8 V/V
# R TOL = 1%
# C TOL = 5%
w_s = 2*np.pi*f_s
w_p = 2*np.pi*f_p
print("ws =", w_s)
print("wp =", w_p)
omega_ratio = w_s/w_p

epsilon = (10**(A_max_db / 10) - 1)**0.5
print("epsilon =", epsilon)

A_ws = 10 * np.log10(1 + epsilon**2 * (omega_ratio**(2 * N)))
print("for N = 11: A(ws) =", A_ws)

# Step 1: Epsilon and radius
epsilon = (10**(A_max_db / 10) - 1)**0.5
radius = w_p * ((1 / epsilon)**(1 / N))
print(f"Pole circle radius = {radius:.2f} rad/s")

# Step 2: Pole angles (starting π/2 + π/2N from jw axis)
theta_0 = np.pi/2 + np.pi/(2*N)
dtheta = np.pi / N

# Step 3: Generate pole angles
poles = []
for k in range(N):
    theta_k = theta_0 + k * dtheta
    pole = radius * np.exp(1j * theta_k)
    if pole.real < 0:  # only left-half plane
        poles.append(pole)

# Step 4: Display results
print("\nButterworth poles (11th order):")
for i, p in enumerate(poles, 1):
    print(f"p{i}: {p.real:.2f} {'+' if p.imag >= 0 else '-'} {abs(p.imag):.2f}j")

# Known natural frequency for all stages
w0 = 93536.64  # rad/s

# Real parts of complex poles (already derived)
real_parts = [13311.65, 38856.53, 61253.48, 78688.03, 89747.75]

# Compute Q values from real parts
for i, real in enumerate(real_parts, 1):
    Q = w0 / (2 * real)
    print(f"Q{i} = {Q:.3f}   (w₀ = {w0} rad/s)")

# Fixed parameters
C2 = 10e-9             # 10 nF
Q_values = [3.5133, 1.2036, 0.7635, 0.5943, 0.5211]  # Q for stages 1 to 5

print("Component values for Stages 1–5 (Sallen-Key,Second-order K = 1):")
for i, Q in enumerate(Q_values, 1):
    C4 = C2 / (4 * Q**2)
    R = 1 / (w0 * np.sqrt(C2 * C4))

    print(f"\nStage {i}:")
    print(f"  Q  = {Q:.3f}")
    print(f"  C2 = {C2 * 1e9:.1f} nF")
    print(f"  C4 = {C4 * 1e9:.2f} nF")
    print(f"  R  = {R / 1e3:.2f} kΩ")

# Stage 6 (first order) 
K = 8
w0 = 93536.64     
R1 = 10e3         
R2 = 80e3           
C = 1/ (w0* R2)   

# Smart S&K formulas
C4 = C2 / (4 * Q**2)
R = 1 / (w0 * np.sqrt(C2 * C4))

print("Stage 6:")
print(f"  C = {C * 1e9:.2f} nF")
print(f"  R1  = {R1 / 1e3:.2f} kΩ")
print(f"  R2 = {R2 / 1e3:.1f} kΩ")
print(f"  K = {K}")
```
