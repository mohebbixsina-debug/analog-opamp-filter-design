```python
#specs
VDD = 2.5            # V
VSS = -2.5           # V
SR = 5.1e6           #  +5.1/-5.1 V/us (original spec: 5 V/us)
CL = 5e-12           # 5 pF
Vin_cm_max = 2.1     # V
Vin_cm_min = -1.3    # V
Vout_max = 2.2       # V
Vout_min = -2.2      # V
GBW = 5.5e6          # 5,5 MHz (original spec: 5 MHz)
PM = 60
Gain_dB = 80         # dB
Sn_sqrt = 30e-9
####################
GB = 2 * np.pi * GBW # Hz (5 MHz)
Bp = 41.5e-6        # hole mobility in m^2/Vs 
# Select process parameters
Bn = 182e-6  # NMOS mobility (m^2/Vs), 350 cm²/Vs
Cox = 3.6e-3    # Oxide capacitance (F/m²)
un = Bn / Cox
up = Bp / Cox
PM_rad = np.radians(PM)    
T = 300              # K (temperature)
k = 1.38e-23  # Boltzmann constant (J/K)
VTN = 0.71           # V (NMOS threshold voltage)
VTP = 0.901          # V (PMOS threshold voltage)
Sn_sqrt = 30e-9      # V/sqrt(Hz)
Sn = Sn_sqrt ** 2    # V^2/Hz
Gain_linear = 10 ** (Gain_dB / 20)

####################
#Analytical Formulas:

# Overdrive voltage of transistor M3
Vov_3 = VDD - VTP + VTN - Vin_cm_max

# Compute compensation capacitor Cc
Cc = (16 * k * T * (1 + (SR / (GB * Vov_3)))) / (3 * GB * Sn)

# Compute I7 from slew rate and total capacitance (Cc + CL)
I7 = SR * (Cc + CL)

# M6
I6 = I7
Vov_6 = VDD - Vout_max 
L6 = np.sqrt((1.5 * up * Vov_6 * Cc) / (GB * (Cc + CL) * np.tan(PM_rad)))
S6 = (2 * I6) / (Bp * Vov_6**2)
W6 = S6 * L6

I5 = SR * Cc

I1 = I5 / 2
I2 = I5 / 2
I3 = I5 / 2
I4 = I5 / 2
Vov_1 = SR / GB
S1 = (2 * I1) / (Bn * Vov_1**2)

S2 = S1

Vov_5 = Vin_cm_min - VTN - VSS - Vov_1
S5 = (2 * I5) / (Bn * Vov_5**2)

Vov_7 = Vov_5
S7 = ((Cc + CL)/Cc)*S5
S3 = (S6 * S5 ) / (2 * S7)
S4 = S3
S9 = (Cc/(Cc + CL))*S6

S10 = S9
S11 = S9
S12 = S9
S13 = S9

I8 = I5
Vov_8 = Vov_5
S8 = S5

RB = Vov_8 / (2*I8)

Vov_14 = Vov_8 / 2
I14 = I8 
S14 = (2 * I14) / (Bn * Vov_14**2)

print(f"I1 = {I1:.4e} A")
print(f"I2 = {I2:.4e} A")
print(f"I3 = {I3:.4e} A")
print(f"I4 = {I4:.4e} A")
print(f"I5 = {I5:.4e} A")
print(f"I6 = {I6:.4e} A")
print(f"I7 = {I7:.4e} A")
print(f"I8 = {I8:.4e} A")

print("Vov_3 = {:.4f} V".format(Vov_3))
print("Vov_6 = {:.4f} V".format(Vov_6))
print("Vov_5 = {:.4f} V".format(Vov_5))
print("Vov_7 = {:.4f} V".format(Vov_7))
print("Vov_8 = {:.4f} V".format(Vov_8))
print("Vov_1 = {:.4f} V".format(Vov_1))
print("Vov_14 = {:.4f} V".format(Vov_14))

print("Cc =", Cc, "F")
print("RB =", RB, "ohm")
for i in range(1, 14):
    value = globals()[f"S{i}"]
    print(f"S{i} (W/L){i} = {value:.4f}")
    
print(f"L6 = {L6:.4e} m")
print(f"W6 = {W6:.4e} m")
```
